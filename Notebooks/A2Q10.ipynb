```python
# your code
import numpy as np
from astropy.coordinates import SkyCoord
from astropy import units as u

# Define the coordinates of the stars in the Square of Pegasus
star_names = ["Alpheratz", "Scheat", "Markab", "Algenib"]
coordinates = SkyCoord(
    ra=["00h08m23.258s", "23h03m46s", "23h04m45.65345s", "00h13m14s"],
    dec=["+29d05m25.55s", "+28d04m58s", "+15d12m18.9617s", "+15d11m00s"],
    unit=(u.hourangle, u.deg),
    frame="icrs"
)

# Define the sides of the square by index
sides = [(0, 1), (1, 2), (2, 3), (3, 0)]

# Compute angular distances
distances = [
    (star_names[i], star_names[j], coordinates[i].separation(coordinates[j]).deg)
    for i, j in sides
]

# Print results
for star1, star2, distance in distances:
    print(f"Angular distance between {star1} and {star2}: {distance:.2f} degrees")
```
