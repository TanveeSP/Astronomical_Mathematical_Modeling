```python
# Your code
from astropy.constants import G, M_earth, R_earth
import numpy as np

# Constants from Astropy
G = G.value              # Gravitational constant (m^3 kg^-1 s^-2)
M = M_earth.value        # Mass of the Earth (kg)
R_E = R_earth.value      # Radius of Earth (m)
T = 86164.1              # Sidereal day in seconds (s)

# Calculate GM
GM = G * M

# Calculate r^3 using the formula GM/r^3 = 4π^2/T^2
r_cubed = (GM * T**2) / (4 * np.pi**2)

# Calculate r (distance from Earth's center to satellite)
r = np.cbrt(r_cubed)

# Convert to units of Earth's radius
r_in_earth_radii = r / R_E

# Output the result
print(f"Distance of geostationary satellite from Earth's center: {r_in_earth_radii:.2f} Earth radii")

```
