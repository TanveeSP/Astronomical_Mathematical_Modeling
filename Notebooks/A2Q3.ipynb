```python
# Your code
# Given
moon_rate_cm_per_year = 3.8  # cm/year
moon_distance_km = 384000  # km
hubble_rate_km_per_s_per_Mpc = 72  # km/s per Mpc

# Conversion factors
cm_to_km = 0.00001  # 1 cm = 0.00001 km
seconds_per_year = 365.25 * 24 * 3600  # seconds in a year
km_per_Mpc = 3.0857e19  # 1 Mpc = 3.0857 * 10^19 km

# 1: Converting Moon's rate from cm/year to km/year
moon_rate_km_per_year = moon_rate_cm_per_year * cm_to_km

#  2: Converting km/year to km/s
moon_rate_km_per_s = moon_rate_km_per_year / seconds_per_year

#  3: Converting the distance from km to Mpc
moon_distance_Mpc = moon_distance_km / km_per_Mpc

#  4: Calculating the rate in km/s per Mpc
moon_rate_km_per_s_per_Mpc = moon_rate_km_per_s / moon_distance_Mpc

#  5: Comparing with Hubble's law
comparison = moon_rate_km_per_s_per_Mpc / hubble_rate_km_per_s_per_Mpc

# Output the results
print(f"Rate of moving away (Moon): {moon_rate_km_per_s_per_Mpc} km/s per Mpc")
print(f"Comparison with Hubble's law: {comparison}")



```
