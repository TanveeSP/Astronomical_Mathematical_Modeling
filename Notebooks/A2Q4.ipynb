```python
# Your code
from astropy import units as u
from astropy.constants import G, M_sun

# Given values
v = 300 * 1000 * u.m / u.s  # speed in m/s (converted from km/s)
r = 0.015 * 3.086e16 * u.m  # distance in meters (converted from parsecs)

# Using the formula v = √(GM/r), solve for M
M = (v**2 * r) / G

# Convert mass to solar masses (M_sun)
M_in_solar_masses = M / M_sun

# Output the mass of the black hole in solar masses
print(f"Mass of the supermassive black hole: {M_in_solar_masses.value} M_sun")



```
