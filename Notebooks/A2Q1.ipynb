```python
# Your code
from astropy.time import Time, TimeDelta

# length of a lunar month = (29.53 days)
lunar_month_duration = TimeDelta(29.53, format='jd')  # in Julian days

# length of a year (365 days)
year_duration = TimeDelta(365, format='jd')  # in Julian days

# Calculate the number of lunar months in a year
lunar_months_in_a_year = year_duration.jd / lunar_month_duration.jd


print(f"The number of lunar months in a year is {lunar_months_in_a_year:.2f}")



```
