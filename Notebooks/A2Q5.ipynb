```python
# Your code
from astropy.constants import G, M_sun
from astropy.units import pc, m, s

# Given values
acceleration = 3 * (m / s**2)  # Acceleration in m/s^2
distance = 0.01 * pc  # Distance in parsecs

# Calculate the mass of the black hole using a = GM/r^2
mass_black_hole = (acceleration * distance**2 / G).to(M_sun)

print(f"The mass of the supermassive black hole is approximately {mass_black_hole:.2e}.")
```
