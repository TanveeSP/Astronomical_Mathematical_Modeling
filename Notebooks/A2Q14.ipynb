```python
# Your code

import numpy as np
import matplotlib.pyplot as plt

L0 = 1e9
k = 0.693
time_days = np.linspace(0, 50, 500)

#Luminosity decay function

L_t = L0 * np.exp(-k * time_days)

# Calculating the time when luminosity drops to 10^6 L_sun
L_target = 1e6
t_target = np.log(L0 / L_target) / k


plt.figure(figsize=(6,4))
plt.plot(time_days, L_t, label="Luminosity Decay", color="blue")
plt.axvline(t_target, color="red", linestyle="--", label=f"Time to reach {L_target:.0e} L_sun")
plt.title("Luminosity Decay of a Type Ia Supernova")
plt.xlabel("Time (days)")
plt.ylabel("Luminosity (L_sun)")
plt.legend()
plt.grid(True)
plt.show()

# Print the time to reach the target luminosity
print(f"It takes {t_target:.2f} days for the luminosity to drop to {L_target:.0e} L_sun.")






```
