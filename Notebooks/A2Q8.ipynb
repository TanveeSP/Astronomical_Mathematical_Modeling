```python
# Your code
from geopy.distance import geodesic

# Coordinates of Mumbai and Mombasa
mumbai_coords = (19.0760, 72.8777)  # Lat, long mumbai
mombasa_coords = (-4.0435, 39.6682)  # Lat, Long Mombasa

# distance in kilometers
distance_km = geodesic(mumbai_coords, mombasa_coords).kilometers

# Radius of Earth in kilometers
R_earth_km = 6371

# Convert the distance into R_earth units
distance_R_earth = distance_km / R_earth_km

print(f"Distance in terms of Earth's radius: {distance_R_earth} R_earth")
print(f"Distance between Mumbai and Mombasa: {distance_km} km")




```
