```python
# Your code
import numpy as np
from astropy import units as u



L_lamp = 100 * u.W  # Lamp's luminosity in watts
solar_flux = 1361 * u.W / u.m**2  # Solar flux at Earth in W/m^2
vega_flux = 3.6e3 * u.W / u.m**2  # Vega flux in W/m^2

def calculate_distance(L, flux):
    return np.sqrt(L / (4 * np.pi * flux))

# Distance for the lamp to match the solar flux at Earth
distance_sun = calculate_distance(L_lamp, solar_flux)

# Distance for the lamp to match the flux of Vega
distance_vega = calculate_distance(L_lamp, vega_flux)

# Output distances in meters
print(f"Distance for lamp to match the solar flux at Earth: {distance_sun:.2f}")
print(f"Distance for lamp to match the flux of Vega: {distance_vega:.2f}")



```
