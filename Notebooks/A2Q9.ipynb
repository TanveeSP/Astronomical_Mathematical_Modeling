```python
# Your code
import numpy as np
from astropy.coordinates import SkyCoord
import astropy.units as u

# coordinates of the stars (RA,dec)(hr,deg)

fomalhaut = SkyCoord(ra=22.960417 * u.deg, dec=-29.6225 * u.deg, frame='icrs')
canopus = SkyCoord(ra=96.157917 * u.deg, dec=-52.695556 * u.deg, frame='icrs')
achernar = SkyCoord(ra=24.928333 * u.deg, dec=-57.236667 * u.deg, frame='icrs')

# Calculate the angular separations between the stars
sep_fomalhaut_canopus = fomalhaut.separation(canopus)
sep_fomalhaut_achernar = fomalhaut.separation(achernar)
sep_canopus_achernar = canopus.separation(achernar)

# Print the results in degrees
print(f"Angular separation between Fomalhaut and Canopus: {sep_fomalhaut_canopus.deg:.2f} degrees")
print(f"Angular separation between Fomalhaut and Achernar: {sep_fomalhaut_achernar.deg:.2f} degrees")
print(f"Angular separation between Canopus and Achernar: {sep_canopus_achernar.deg:.2f} degrees")



```
